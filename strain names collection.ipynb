```python
import json

with open('..\\dissertation DLC content\\big_list_files\\names_report.jsonl', 'r') as f:
    strain_names = f.readlines()
```

```python
strains = []

with open('list_files\\ncbi_ids.json', 'r') as f:
    ncbi_ids = json.load(f)

ncbi_ids = set(ncbi_ids)

for strain in strain_names:
    line = json.loads(strain).get('taxonomy')
    names = line.get('currentScientificName', {}).get('name')
    ncbi_id = line.get('taxId')
    if ncbi_id not in ncbi_ids:
        ncbi_ids.add(ncbi_id)
    if not names:
        continue
    names = names.split(' ')
    if not names[-1].isalpha() and names[-1] != 'sp.' and names[-1][-1] != ')' and names[-1][0] != '(':
        strains.append(names[-1])

print(len(strains))

with open('list_files\\strain_names.json', 'w', encoding='utf-8') as f:
    json.dump(strains, f)

with open('list_files\\ncbi_ids.json', 'w') as f:
    json.dump(list(ncbi_ids), f)
```
